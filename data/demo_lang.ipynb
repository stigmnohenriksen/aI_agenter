```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
llm = ChatOllama(
    model="llama3.1",
    temperature=0.7,
)
```

```python
system_template = """
Du er en tur planlegger ekspert. Hjelp meg å planlegge en tur til {destination}
Her er mine {preferences} for turen:
"""

prompt_template = ChatPromptTemplate.from_messages(
    [
        ("system", system_template),  # system rolle 
        ("user", " Hva skal jeg gjøre på min ferie {destination}?"),  # brukerrolle
    ]
)
```

```python
parser = StrOutputParser()
```

```python
trip_planner_chain = prompt_template | llm | parser
```

```python
def plan_trip(destination: str, preferences: list):
     input_data = {
         "destination": destination,
         "preferences": preferences
     }
     return trip_planner_chain.invoke(input_data)
```

```python
from IPython.display import display, Markdown
def render_markdown(md_string):
    display(Markdown(md_string))
```

```python
resultat = plan_trip("Paris", ["kunst", "mat", "historie"])
```
