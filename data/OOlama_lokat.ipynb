```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# 1. Initialiser den lokale modellen
# 'model' må matche navnet på modellen du har i Ollama (sjekk 'ollama list')
llm = ChatOllama(
    model="llama3.1",
    temperature=0.7,
)

# 2. Lag en template
prompt = ChatPromptTemplate.from_messages([
    ("system", "Du er en erfaren programmerer som forklarer konsepter enkelt."),
    ("user", "{input}")
])

# 3. Sett opp kjeden (Chain)
# Vi legger til en OutputParser for å få ren tekst ut i steden for et objekt
chain = prompt | llm | StrOutputParser()

# 4. Kjør og skriv ut svar
response = chain.invoke({"input": "Hva er fordelen med å bruke React sammen med et Python API?"})

print(response)
```

## Kjører med Mistral som model

```python
from langchain_ollama import ChatOllama
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate

# 1. Initialiser modellen med nøyaktig navn fra din 'ollama list'
llm = ChatOllama(
    model="mistral:latest",
    temperature=0,  # 0 gjør svarene mer forutsigbare og faktiske
)

# 2. Lag en enkel spørring
prompt = ChatPromptTemplate.from_template("Fortell meg tre fordeler med å bruke {modellnavn} lokalt.")

# 3. Sett opp kjeden
chain = prompt | llm | StrOutputParser()

# 4. Kjør testen
try:
    print("Sender spørsmål til Mistral...\n")
    response = chain.invoke({"modellnavn": "Mistral"})
    print("-" * 30)
    print(response)
    print("-" * 30)
except Exception as e:
    print(f"Oops, noe gikk galt: {e}")
```
